**Lane: Lane 3 — Structured Content Archetype Clustering.**

I'm choosing this lane because I want to understand what *kinds* of pages exist across the content inventory before deciding where review effort should go, rather than jumping straight to a single ranked to-do list. With over half a million content items, no reviewer can eyeball them individually — clustering surfaces natural groupings of pages that behave alike (e.g. strong performers, stale-but-visible pages, low-demand pages), so a strategist can apply one action to a whole group instead of one page at a time. I'm clustering on **observable metrics and buckets only** — this dataset has no article text, so this is behavioral clustering, not semantic/topical clustering, and I won't describe it as the latter.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Question:** What performance archetypes exist across the content inventory, and what does each archetype suggest about how a page should be treated?

**Unit of analysis:** one content item (page).

**Decision improved:** which treatment — protect, improve, rewrite, merge, prune, or monitor — fits a page, based on the archetype it falls into.

**Action someone takes:** a content strategist reviews a cluster's profile (not each page individually) and applies a bulk action to the pages inside it, or uses cluster membership to prioritize which group to dig into first.

**Cost of a wrong call:** two failure directions matter differently here. Grouping a page with real hidden demand into a 'prune' archetype risks losing traffic that was recoverable. Grouping a genuinely weak page into a 'protect' archetype wastes review capacity defending something not worth defending. For this project I'm treating the first error (wrongly pruning a page with real demand) as more costly, since traffic loss is harder to reverse than a wasted review.

**Why data/ML helps at all:** at this scale (519,606 content items), manual grouping isn't feasible, and simple single-metric rules (e.g. 'sort by impressions') miss pages that are unusual across *combinations* of signals — clustering can surface a group like 'high impressions, low engagement' that a single-column sort would never isolate.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Apply the guide's filter rule: impressions_90d > 0, content_age_days >= 90
df = df[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]
df = df.drop_duplicates(subset="content_id")

print("Rows after filtering:", len(df))

# Number 1: spread of engagement rate — is there enough variety to cluster on?
print("\nEngagement rate summary:")
print(df["engagement_rate"].describe())

# Number 2: spread of average position — do pages sit at very different visibility levels?
print("\nAverage position summary:")
print(df["avg_position"].describe())

# Number 3: spread of word count — structural variety worth clustering on
print("\nWord count summary:")
print(df["word_count"].describe())
```
